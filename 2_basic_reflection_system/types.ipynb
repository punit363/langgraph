```python
from pydantic import BaseModel , Field
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(model="gpt-4o")

class Country(BaseModel):
    """Information about a country"""

    # Parameters
    name:str=Field(description="name of the country") # property
    language:str=Field(description="language of the country") # property
    capital:str=Field(description="capital city of the country") # property

# with_structured_output method also validates the output by llm and throws an error if it does not match
structured_llm = llm.with_structured_output(Country)

structured_llm.invoke("Tell me about India")
```

```python
from typing_extensions import TypedDict, Annotated
from typing import Optional

class Joke(TypedDict):
    """"Joke to tell user"""

    setup: Annotated[str,...,"The setup of a funny joke"]

    punchline:Annotated[str,...,"The punchline of the joke"]

    rating:Annotated[Optional[int],None,"The rating of the joke on a scale of 1 to 10"]

structured_llm = llm.with_structured_output(Joke)

structured_llm.invoke("Tell me about India")
```

```python
json_schema={
    "title":"joke",
    "description":"Joke to tell user",
    "type":"object",
    "properties":{
        "setup":{
            "type":"string",
            "description":"The setup of the joke"
        },
        "setup":{
            "type":"string",
            "description":"The punchline of the joke"
        },
        "setup":{
            "type":"string",
            "description":"Rate the joke on a scale of 1 to 10",
            "default":None
        },
    },
    "required":["setup","punchline"]
}

structured_llm = llm.with_structured_output(Joke)

structured_llm.invoke("Tell me about India")
```
